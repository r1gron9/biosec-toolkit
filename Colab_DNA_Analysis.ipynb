### Import Libraries

```python
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt
from difflib import SequenceMatcher
import warnings
warnings.filterwarnings('ignore')

print("✓ All libraries imported!")
```

### Load Functions

```python
def generate_random_sequence(length=300):
    """Generate a random DNA sequence."""
    nucleotides = ['A', 'T', 'G', 'C']
    return ''.join(random.choice(nucleotides) for _ in range(length))

def inject_attata_motif(sequence, motif='ATTATA', position=None):
    """Inject ATTATA promoter motif into a DNA sequence."""
    seq_length = len(sequence)
    if position is None:
        position = random.randint(0, max(0, seq_length - len(motif)))
    modified_seq = sequence[:position] + motif + sequence[position + len(motif):]
    return modified_seq, position

def search_motif_occurrences(sequence, motif='ATTATA'):
    """Search for all occurrences of a motif in a sequence."""
    count = 0
    positions = []
    for i in range(len(sequence) - len(motif) + 1):
        if sequence[i:i+len(motif)] == motif:
            count += 1
            positions.append(i)
    return {'count': count, 'positions': positions}

def compare_sequences(seq1, seq2):
    """Compare two DNA sequences."""
    matcher = SequenceMatcher(None, seq1, seq2)
    similarity = matcher.ratio() * 100
    differences = sum(1 for a, b in zip(seq1, seq2) if a != b)
    return {'similarity': round(similarity, 2), 'differences': differences}

def analyze_nucleotide_composition(sequence):
    """Analyze nucleotide composition."""
    seq_upper = sequence.upper()
    total = len(seq_upper)
    composition = {}
    for nucleotide in ['A', 'T', 'G', 'C']:
        count = seq_upper.count(nucleotide)
        composition[nucleotide] = round((count / total) * 100, 2)
    gc_content = composition['G'] + composition['C']
    composition['GC_content'] = round(gc_content, 2)
    return composition

print("✓ All functions loaded!")
```

# Optimizing a Single Sequence

### Enter DNA Sequence

```python
# 📝 EDIT THIS CELL - Enter your DNA sequence below
dna_sequence = "ATGATGCTCGAATTGTACGACAAATACGCCTACTACCTGGTCTGCTGACCATCCGCATCGATGTGGTCGCCGTGTACTACATCATCGACATGGACCCGGTGAAGATGCGCGTCATGGCCGACGTGACCTACGAAGTCAACGCCGTCGACCCGCTGTACGTGGTCGACTACGCCGTCTGGGTCGAGGGCAAAGCCGTCGAAATCCCGCTGCTGACCATGGTCTACACCTACGACAAAAACACCACGCTGCTGGATCCGCTGACCTACTACGAGAAGATCTACGACAAACTGCTGCACATGATCAAACCGCTGCGCAACATCATCTACCCGGTCGCCGAAATCGTCGCCAAAGACCTGTACGTCGACTACTATGCCGTCTACGACGTCGATGTCCGCGCCGACGTCGCCGTCAAACTGACCCACGAAGGCGAAATCGTCGCCAACGACGTCGAA"

# Optional: Custom motif
motif = "ATTATA"

print(f"Sequence Length: {len(dna_sequence)} bp")
print(f"Motif: {motif}")
print(f"First 70 bp: {dna_sequence[:70]}...")
```

### Run Optimization!

```python
# Analyze original sequence
original_comp = analyze_nucleotide_composition(dna_sequence)
original_motifs = search_motif_occurrences(dna_sequence, motif)

print("📊 ORIGINAL SEQUENCE ANALYSIS")
print(f"  Length: {len(dna_sequence)} bp")
print(f"  A: {original_comp['A']}% | T: {original_comp['T']}% | G: {original_comp['G']}% | C: {original_comp['C']}%")
print(f"  GC Content: {original_comp['GC_content']}%")
print(f"  {motif} Motifs: {original_motifs['count']}")
print()

# Inject motif
optimized_seq, position = inject_attata_motif(dna_sequence, motif)

# Analyze optimized sequence
optimized_comp = analyze_nucleotide_composition(optimized_seq)
optimized_motifs = search_motif_occurrences(optimized_seq, motif)

print("✨ OPTIMIZED SEQUENCE ANALYSIS")
print(f"  Length: {len(optimized_seq)} bp")
print(f"  A: {optimized_comp['A']}% | T: {optimized_comp['T']}% | G: {optimized_comp['G']}% | C: {optimized_comp['C']}%")
print(f"  GC Content: {optimized_comp['GC_content']}%")
print(f"  {motif} Motifs: {optimized_motifs['count']}")
print(f"  Injection Position: {position}")
print()

# Comparison
comparison = compare_sequences(dna_sequence, optimized_seq)
print("🔄 COMPARISON")
print(f"  Similarity: {comparison['similarity']}%")
print(f"  Differences: {comparison['differences']} bp")
```

# Batch Processing Sequences

### Enter Dataset Parameters

```python
# 📝 EDIT THIS CELL - Customize your batch processing
num_sequences = 10      # Number of sequences to generate
seq_length = 500        # Length of each sequence in bp
output_filename = "batch_results.csv"  # Output filename

print(f"Batch Processing Configuration:")
print(f"  Sequences to generate: {num_sequences}")
print(f"  Length per sequence: {seq_length} bp")
print(f"  Output file: {output_filename}")
```

### Run Batch Processing!

```python
# Generate sequences
sequences_list = []
for i in range(num_sequences):
    seq = generate_random_sequence(seq_length)
    sequences_list.append({
        'ID': f'seq_{i+1}',
        'sequence': seq,
        'length': len(seq)
    })

df = pd.DataFrame(sequences_list)
print(f"✓ Generated {len(df)} sequences")
print(f"\nSequence Summary:")
print(df.head())
```

```python
# Inject motif into all sequences
injected_sequences = []
for idx, row in df.iterrows():
    modified_seq, pos = inject_attata_motif(row['sequence'])
    injected_sequences.append({
        'ID': row['ID'],
        'original_sequence': row['sequence'],
        'injected_sequence': modified_seq,
        'injection_position': pos,
        'original_length': len(row['sequence']),
        'modified_length': len(modified_seq)
    })

injected_df = pd.DataFrame(injected_sequences)
print(f"✓ Injected ATTATA motif into {len(injected_df)} sequences")
print(f"\n{injected_df[['ID', 'injection_position', 'original_length', 'modified_length']]}")
```

```python
# Analyze motifs in all sequences
motif_results = []
for idx, row in injected_df.iterrows():
    analysis = search_motif_occurrences(row['injected_sequence'])
    motif_results.append({
        'ID': row['ID'],
        'ATTATA_count': analysis['count'],
        'positions': str(analysis['positions'])
    })

motif_df = pd.DataFrame(motif_results)
print(f"✓ Analyzed ATTATA motifs")
print(f"\n{motif_df}")
print(f"\nSummary:")
print(f"  Total Sequences: {len(motif_df)}")
print(f"  Total ATTATA: {motif_df['ATTATA_count'].sum()}")
print(f"  Avg Motifs/Seq: {motif_df['ATTATA_count'].mean():.2f}")
```

```python
# Analyze nucleotide composition
composition_data = []
for idx, row in df.iterrows():
    comp = analyze_nucleotide_composition(row['sequence'])
    composition_data.append({
        'ID': row['ID'],
        'A_percent': comp['A'],
        'T_percent': comp['T'],
        'G_percent': comp['G'],
        'C_percent': comp['C'],
        'GC_content': comp['GC_content']
    })

comp_df = pd.DataFrame(composition_data)
print(f"✓ Analyzed nucleotide composition")
print(f"\n{comp_df}")
print(f"\nAverage GC Content: {comp_df['GC_content'].mean():.2f}%")
```

```python
# Save all results to CSV
injected_df.to_csv('injected_sequences.csv', index=False)
motif_df.to_csv('motif_analysis.csv', index=False)
comp_df.to_csv('nucleotide_composition.csv', index=False)

print("✓ Files saved:")
print("  - injected_sequences.csv")
print("  - motif_analysis.csv")
print("  - nucleotide_composition.csv")

# Download from Colab
try:
    from google.colab import files
    print("\n📥 Downloading...")
    files.download('injected_sequences.csv')
    files.download('motif_analysis.csv')
    files.download('nucleotide_composition.csv')
except ImportError:
    print("\n(Not running in Colab)")
```
