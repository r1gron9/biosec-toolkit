```python
#@markdown ## Setup (Run Once)

!pip install -q pandas numpy

import pandas as pd
import numpy as np
import random
from difflib import SequenceMatcher
import warnings
warnings.filterwarnings('ignore')

def generate_random_sequence(length=300):
    nucleotides = ['A', 'T', 'G', 'C']
    return ''.join(random.choice(nucleotides) for _ in range(length))

def inject_attata_motif(sequence, motif='ATTATA', position=None):
    seq_length = len(sequence)
    if position is None:
        position = random.randint(0, max(0, seq_length - len(motif)))
    modified_seq = sequence[:position] + motif + sequence[position + len(motif):]
    return modified_seq, position

def search_motif_occurrences(sequence, motif='ATTATA'):
    count = 0
    positions = []
    for i in range(len(sequence) - len(motif) + 1):
        if sequence[i:i+len(motif)] == motif:
            count += 1
            positions.append(i)
    return {'count': count, 'positions': positions}

def compare_sequences(seq1, seq2):
    matcher = SequenceMatcher(None, seq1, seq2)
    similarity = matcher.ratio() * 100
    differences = sum(1 for a, b in zip(seq1, seq2) if a != b)
    return {'similarity': round(similarity, 2), 'differences': differences}

def analyze_nucleotide_composition(sequence):
    seq_upper = sequence.upper()
    total = len(seq_upper)
    composition = {}
    for nucleotide in ['A', 'T', 'G', 'C']:
        count = seq_upper.count(nucleotide)
        composition[nucleotide] = round((count / total) * 100, 2)
    gc_content = composition['G'] + composition['C']
    composition['GC_content'] = round(gc_content, 2)
    return composition

print("Ready")
```

## Optimize Your Sequence

```python
#@markdown Paste your DNA sequence and click play

dna_sequence = "ATGATGCTCGAATTGTACGACAAATACGCCTACTACCTGGTCTGCTGACCATCCGCATCGATGTGGTCGCCGTGTACTACATCATCGACATGGACCCGGTGAAGATGCGCGTCATGGCCGACGTGACCTACGAAGTCAACGCCGTCGACCCGCTGTACGTGGTCGACTACGCCGTCTGGGTCGAGGGCAAAGCCGTCGAAATCCCGCTGCTGACCATGGTCTACACCTACGACAAAAACACCACGCTGCTGGATCCGCTGACCTACTACGAGAAGATCTACGACAAACTGCTGCACATGATCAAACCGCTGCGCAACATCATCTACCCGGTCGCCGAAATCGTCGCCAAAGACCTGTACGTCGACTACTATGCCGTCTACGACGTCGATGTCCGCGCCGACGTCGCCGTCAAACTGACCCACGAAGGCGAAATCGTCGCCAACGACGTCGAA" #@param {type:"string"}
motif = "ATTATA" #@param {type:"string"}

original_comp = analyze_nucleotide_composition(dna_sequence)
original_motifs = search_motif_occurrences(dna_sequence, motif)
optimized_seq, position = inject_attata_motif(dna_sequence, motif)
optimized_comp = analyze_nucleotide_composition(optimized_seq)
optimized_motifs = search_motif_occurrences(optimized_seq, motif)
comparison = compare_sequences(dna_sequence, optimized_seq)

print("\n" + "="*50)
print("ORIGINAL")
print("="*50)
print(f"Length: {len(dna_sequence)} bp")
print(f"Composition: A={original_comp['A']}%, T={original_comp['T']}%, G={original_comp['G']}%, C={original_comp['C']}%")
print(f"GC Content: {original_comp['GC_content']}%")
print(f"ATTATA Count: {original_motifs['count']}")

print("\n" + "="*50)
print("OPTIMIZED")
print("="*50)
print(f"Length: {len(optimized_seq)} bp")
print(f"Composition: A={optimized_comp['A']}%, T={optimized_comp['T']}%, G={optimized_comp['G']}%, C={optimized_comp['C']}%")
print(f"GC Content: {optimized_comp['GC_content']}%")
print(f"ATTATA Count: {optimized_motifs['count']}")
print(f"Motif Position: {position}")

print("\n" + "="*50)
print("COMPARISON")
print("="*50)
print(f"Similarity: {comparison['similarity']}%")
print(f"Changes: {comparison['differences']} bp")
print("="*50)
```

## Compare Two Sequences

```python
#@markdown Enter two sequences to compare

sequence1 = "ATGATGCTCGAATTGTACGACAAATACGCCTACTACCTGGTCTGCTGACCATCCGCATCGATGTGGTCGCCGTGTACTACATCATCGACATGGACCCG" #@param {type:"string"}
sequence2 = "ATGATGCTCGAATTGTACGACAAATACGCCTACTACCTGGTCTGCTGACCATCCGCATCGATGTGGTCGCCGTGTACTACATCATCGACATGGACCCG" #@param {type:"string"}

comparison_result = compare_sequences(sequence1, sequence2)
seq1_comp = analyze_nucleotide_composition(sequence1)
seq2_comp = analyze_nucleotide_composition(sequence2)

print("\n" + "="*50)
print("SEQUENCE 1")
print("="*50)
print(f"Length: {len(sequence1)} bp")
print(f"Composition: A={seq1_comp['A']}%, T={seq1_comp['T']}%, G={seq1_comp['G']}%, C={seq1_comp['C']}%")
print(f"GC Content: {seq1_comp['GC_content']}%")

print("\n" + "="*50)
print("SEQUENCE 2")
print("="*50)
print(f"Length: {len(sequence2)} bp")
print(f"Composition: A={seq2_comp['A']}%, T={seq2_comp['T']}%, G={seq2_comp['G']}%, C={seq2_comp['C']}%")
print(f"GC Content: {seq2_comp['GC_content']}%")

print("\n" + "="*50)
print("COMPARISON")
print("="*50)
print(f"Similarity: {comparison_result['similarity']}%")
print(f"Differences: {comparison_result['differences']} bp")
print("="*50)
```

## Analyze Batch Sequences

```python
#@markdown Generate and analyze multiple sequences

num_sequences = 10 #@param {type:"integer", min:1, max:100}
seq_length = 500 #@param {type:"integer", min:100, max:5000}

sequences_list = []
for i in range(num_sequences):
    seq = generate_random_sequence(seq_length)
    sequences_list.append({
        'ID': f'seq_{i+1}',
        'sequence': seq,
        'length': len(seq)
    })

df = pd.DataFrame(sequences_list)

injected_sequences = []
for idx, row in df.iterrows():
    modified_seq, pos = inject_attata_motif(row['sequence'])
    injected_sequences.append({
        'ID': row['ID'],
        'original_sequence': row['sequence'],
        'injected_sequence': modified_seq,
        'injection_position': pos
    })

injected_df = pd.DataFrame(injected_sequences)

motif_results = []
for idx, row in injected_df.iterrows():
    analysis = search_motif_occurrences(row['injected_sequence'])
    motif_results.append({
        'ID': row['ID'],
        'ATTATA_count': analysis['count']
    })

motif_df = pd.DataFrame(motif_results)

composition_data = []
for idx, row in df.iterrows():
    comp = analyze_nucleotide_composition(row['sequence'])
    composition_data.append({
        'ID': row['ID'],
        'GC_content': comp['GC_content']
    })

comp_df = pd.DataFrame(composition_data)

injected_df.to_csv('injected_sequences.csv', index=False)
motif_df.to_csv('motif_analysis.csv', index=False)
comp_df.to_csv('gc_content_analysis.csv', index=False)

print("\n" + "="*50)
print("BATCH ANALYSIS RESULTS")
print("="*50)
print(f"Sequences Generated: {len(df)}")
print(f"Each Length: {seq_length} bp")
print(f"Total ATTATA Motifs: {motif_df['ATTATA_count'].sum()}")
print(f"Average Motifs per Sequence: {motif_df['ATTATA_count'].mean():.2f}")
print(f"Average GC Content: {comp_df['GC_content'].mean():.2f}%")
print("\nFiles saved:")
print("  - injected_sequences.csv")
print("  - motif_analysis.csv")
print("  - gc_content_analysis.csv")
print("="*50)

try:
    from google.colab import files
    files.download('injected_sequences.csv')
    files.download('motif_analysis.csv')
    files.download('gc_content_analysis.csv')
    print("\nFiles downloaded")
except:
    pass
```
